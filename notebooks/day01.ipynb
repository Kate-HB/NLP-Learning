```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")
```

```python
classifier(
    ["I've been waiting for a HuggingFace course my whole life.", "I hate this so much!"]
)
```

```python
#0样本分类
from transformers import pipeline
classifier = pipeline("zero-shot-classification")
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
#我的尝试
from transformers import pipeline
classifier = pipeline("zero-shot-classification")
classifier(
    "Jack Ma is a great businessman",
    candidate_labels=["Chinese", "American"],
)
```

```python
#文本生成
from transformers import pipeline

generator = pipeline("text-generation",model="openai-community/gpt2")
generator("In this course, we will teach you how to")
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")
generator(
    "In this course, we will teach you how to",
    max_length=30,
    num_return_sequences=2,
)
```

```python
from transformers import pipeline

unmasker = pipeline("fill-mask",model="bert-base-uncased")
unmasker("This course will teach you all about [MASK].",top_k=2)
```

```python
#查看模型对应 Tokenizer 的 mask_token，找到掩码符号
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-uncased"
)

print(tokenizer.mask_token)
print(tokenizer.mask_token_id)
```

```python
#命名实体识别
from transformers import pipeline

ner = pipeline("ner", model="QCRI/bert-base-multilingual-cased-pos-english",aggregation_strategy="simple")
ner("My name is Sylvain and I work at Hugging Face in Brooklyn.")
```

```python
#问题回答
from transformers import pipeline

question_answerer = pipeline("question-answering")
question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Hugging Face in Brooklyn",
)
```

```python
from transformers import pipeline

question_answerer = pipeline("question-answering",model="timpal0l/mdeberta-v3-base-squad2")
question_answerer(
    question="你叫什么名字",
    context="我的名字是罗龙飞"
)
```
